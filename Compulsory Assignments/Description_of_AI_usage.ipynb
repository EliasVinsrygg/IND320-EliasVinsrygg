***How I've used AI in my work:***

I use ChatGPT and Codex actively as development tools when working on the assignment. My goal is to learn to be an efficient developer by using agentic coding to improve productivity without compromising my understanding or the quality of the final product.

My typical workflow is to discuss problems and possible solutions with the AI agent before arriving at a solution that I review and approve. Codex is then used to generate or modify code based on specific instructions. Code that I've written myself is also looked over by Codex, so I can get input on how to improve it. By using Codec directly in the terminal, I can also ask it to push my changes to git or change my Github repository name if I want to do that quickly. But that's also stuff i do on my own normally.

An important principle for me is that I should understand the code and the solutions we develop. I frequently ask the agent to explain problems and suggest some alternatives.
