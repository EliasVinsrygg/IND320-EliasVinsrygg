### Work completed so far

Notebook: The compulsory work began in Jupyter Notebook, where I explored the weekly Norwegian reservoir dataset. After a while, I understood the relation between EL, VASS andre the overall NO-0. I created separate visualisations for EL and VASS areas because the two regional systems are independent from eachothers. I also made a recent-years view to make long time series easier to interpret, because the the full-time plots are kind of noicy.

Streamlit: The app is organised into four pages. I learned how to separate data loading from page code and cache the local CSV file so it is not read repeatedly. The interactive chart uses Plotly, but I tried with Aliair first. A main design decision was to show Norway (NO-0) as a clear national reference while allowing either EL or VASS regions to be displayed, if the user wants to. The graph shows the first month in the dataset, because it was a task requirement. However, I think it makes more desition support to show recent data. I therefore added the "Show recent years" button. 

A useful Streamlit lesson was that interaction state needs deliberate handling. The selected time period must remain unchanged when a regional view is selected, so the app keeps the chart period separately from the slider's temporary UI value
